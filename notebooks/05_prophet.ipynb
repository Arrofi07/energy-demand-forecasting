# Phase 5 — Prophet Baseline

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
from src.data.loader import load_processed
from src.models.prophet_model import fit_prophet, forecast_prophet
from src.evaluation.metrics import evaluate_all
```

```python
df = load_processed()
hourly = df['Global_active_power'].resample('h').mean().dropna()
prophet_df = hourly.reset_index().rename(columns={'datetime': 'ds', 'Global_active_power': 'y'})
```

```python
train = prophet_df[prophet_df['ds'] < '2010-01-01']
test = prophet_df[(prophet_df['ds'] >= '2010-01-01') & (prophet_df['ds'] < '2010-02-01')]

model = fit_prophet(train)
forecast = forecast_prophet(model, test[['ds']])
evaluate_all(test['y'].values, forecast['yhat'].values)
```
